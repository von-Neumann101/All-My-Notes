# 示例：集合的交集（线性时间）

给定两个严格单调递增列表，返回他们相同元素个数（要求时间复杂度为$O(n)$）

```python
def fast_overlap(s, t):
    i, j, count = 0, 0, 0
    while i < len(s) and j < len(t):
        if s[i] == t[j]:
            count, i, j = count + 1, i + 1, j + 1
        elif s[i] < t[j]:
            i += 1
        else:
            j += 1
    return count
```
