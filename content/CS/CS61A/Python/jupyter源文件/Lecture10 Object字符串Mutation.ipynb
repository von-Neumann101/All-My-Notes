# 可变性

## 可变操作

```python
Dict = ['A', 'B', 'C', 'D', 'E']
original_Dict = Dict
```

```python
Dict.pop()#去掉最后一个元素
```

```python
Dict.remove('A')
```

```python
Dict
```

```python
Dict.append('awa')
```

```python
Dict.extend(['qwq', 'QAQ'])
```

```python
Dict[4] = 'QwQ'
```

所有指向相同对象的名字都会受到改变的影响

```python
print(Dict)
print(original_Dict)
```

```python
Dict = {'I': 1, 'V': 5, 'X': 10}
```

```python
Dict['F'] = 4
```

```python
Dict
```
