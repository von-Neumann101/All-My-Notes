# Macros

可以允许定义新的特殊形式

宏是在源代码执行前对其处理的一种操作，先执行宏语句然后产生新表达式，然后再对该表达式求值

```racket
(define (twice expr)
  (list 'begin expr expr)
  )
```

```racket
(twice (print 2)) ;普通定义过程，先对expr求值，然后代入计算
```

```racket
(define-syntax twice
  (syntax-rules ()
    ((_ expr)
     (begin expr expr)))
  )
```

```racket
(twice (print 2))
```

```racket
(define-syntax check
  (syntax-rules ()
    ((_ expr)
     (if expr
         'passed
         (list 'failed: 'expr)
         )
     )
    )
  )
```

```racket
(define x -2)
(check (> x 0))
```

无论如何，常规过程无法得到表达式的原来内容

# for 宏

```racket
(define-syntax for
  (syntax-rules ()
    ((_ var lst expr)
     (map (lambda (var) expr) lst)))
  )
```

```racket
(for x '(2 3 4 5) (* x x))
```
