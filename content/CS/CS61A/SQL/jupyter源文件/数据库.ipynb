# 算术

```SQL
CREATE TABLE lift (
    chair INTEGER,
    single INTEGER,
    couple INTEGER
);
INSERT INTO lift (chair, single, couple) VALUES
    (101, 2, 2),
    (102, 0, 3),
    (103, 4, 1);
```

创建一个数据表

```SQL
SELECT chair, single + 2 * couple AS total FROM lift;
```

```powershell
sqlite> SELECT chair, single + 2 * couple AS total FROM lift;
chair  total
-----  -----
101    6
102    6
103    6
```
